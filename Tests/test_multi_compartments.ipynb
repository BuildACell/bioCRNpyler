```python
from biocrnpyler.mixtures_extract import TxTlExtract
from biocrnpyler.dna_assembly import DNAassembly
from biocrnpyler.dna_part_promoter import ActivatablePromoter
import numpy as np
import pylab as plt
```

```python
# P_activatable = ActivatablePromoter("P_activatable", activator="A", leak=False)
R = DNAassembly(name="R", promoter="strong", rbs="BCD2", protein="protein_R", transcript="mrna_R", initial_concentration=10)
E = TxTlExtract(components=[R], parameter_file="../examples/default_parameters.txt")
CRN = E.compile_crn()
CRN.write_sbml_file("CRN1.xml")
print(CRN)
```
